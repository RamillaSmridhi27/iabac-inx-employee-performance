```python
import joblib
import pandas as pd

# ===== LOAD MODEL =====
model = joblib.load("model.pkl")
columns = joblib.load("feature_columns.pkl")

print("Model Loaded Successfully")

# ===== SAMPLE INPUT DATA =====
input_data = {
    "Age": 30,
    "DistanceFromHome": 5,
    "MonthlyIncome": 5000,
    "JobSatisfaction": 3,
    "WorkLifeBalance": 2
}

# Convert to DataFrame
sample = pd.DataFrame([input_data])

# Ensure correct column order
sample = sample.reindex(columns=columns, fill_value=0)

# ===== PREDICT =====
prediction = model.predict(sample)

print(" Predicted Performance Rating:", prediction[0])
```

```python
Model Loaded Successfully
 Predicted Performance Rating: 3
```

Insights:

The predicted numbers represent the employee’s performance level as classified by the machine learning model.

| Predicted Value | Performance Level          | Interpretation                                        |
| --------------- | -------------------------- | ----------------------------------------------------- |
| **2**           | Low Performer              | Employee is underperforming relative to expectations  |
| **3**           | Medium / Average Performer | Employee is meeting standard performance expectations |
| **4**           | High Performer             | Employee is exceeding performance expectations        |


Deeper Insights:

🔹2 — Low Performance

If the model predicts 2, it suggests:

--> The employee may have low job satisfaction

--> Work-life balance might be weak

--> Income, engagement, or distance factors could be negatively influencing productivity

--> Performance improvement interventions may be required (training, mentoring, workload review)

#Business Insight:
--> HR may need to monitor such employees to reduce attrition risk and improve engagement.

🔹 3 — Medium Performance

If the prediction is 3, it indicates:

--> The employee is performing at an acceptable level

--> Meeting targets but not significantly exceeding them

--> Stable but may benefit from motivation or skill enhancement

#Business Insight:

This group forms the operational backbone of the organization. Development programs can help convert them into high performers.

🔹 4 — High Performance

--> If the model predicts 4, it implies:

--> Strong job satisfaction and work-life balance

--> Likely high productivity and engagement

--> Potential leadership or promotion candidate

#Business Insight:
These employees are high-value assets. Retention strategies and career growth planning are critical.

Overall Model Interpretation

1. Age
